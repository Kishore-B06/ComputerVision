## Let's take a closer look at color spaces

You may have remembered we talked about images being stored in RGB (Red Green Blue) color Spaces. Let's take a look at that in OpenCV.

### First thing to remember about OpenCV's RGB is that it's BGR (I know, this is annoying)

Let's look at the image shape again. The '3L' 

```python
import cv2
import numpy as np

image = cv2.imread('./images/input.jpg')
```

### Let's look at the individual color levels for the first pixel (0,0)

```python
# BGR Values for the first 0,0 pixel
B, G, R = image[0, 0] 
print(B, G, R)
print(image.shape)
```

Let's see what happens when we convert it to grayscale

```python
gray_img = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
print(gray_img.shape)
print(gray_img[10, 50]) 
```

It's now only 2 dimensions. Each pixel coordinate has only one value (previously 3) with a range of 0 to 255

```python
gray_img[0, 0] 
```

### Another useful color space is HSV 
Infact HSV is very useful in color filtering.

```python
import cv2

# Define the range of HSV values
# H: 0 - 180, S: 0 - 255, V: 0 - 255

# Load the image
image = cv2.imread('./images/input.jpg')

# Convert the image to HSV color space
hsv_image = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

# Display the HSV image and its channels
cv2.imshow('HSV image', hsv_image)
cv2.imshow('Hue channel', hsv_image[:, :, 0]) # Hue channel of the HSV image
cv2.imshow('Saturation channel', hsv_image[:, :, 1]) # Saturation channel of the HSV image.
cv2.imshow('Value channel', hsv_image[:, :, 2]) # Value channel of the HSV image.

# Wait for a key press and close all windows
cv2.waitKey()
cv2.destroyAllWindows()
```

### Let's now explore lookng at individual channels in an RGB image

```python
import cv2

image = cv2.imread('./images/input.jpg')

# OpenCV's 'split' function splites the image into each color index
B, G, R = cv2.split(image)

print(B.shape) #Prints the shape (dimensions) of the Blue channel of the image.

#Displays the individual color channels (Red, Green, and Blue) of the image using OpenCV's
cv2.imshow("Red", R) 
cv2.imshow("Green", G)
cv2.imshow("Blue", B)
cv2.waitKey(0)
cv2.destroyAllWindows()

# Let's re-make the original image, 
merged = cv2.merge([B, G, R]) 
cv2.imshow("Merged", merged) 

# Let's amplify the blue color
merged = cv2.merge([B+100, G, R])
cv2.imshow("Merged with Blue Amplified", merged) 

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2              # Import the OpenCV library, providing access to image processing functions.
import numpy as np      # Import the NumPy library, used for numerical operations.

B, G, R = cv2.split(image)  # Split the input image into its individual color channels: Blue (B), Green (G), and Red (R).

# Let's create a matrix of zeros 
# with dimensions of the image h x w  
zeros = np.zeros(image.shape[:2], dtype="uint8")  # Create a zero matrix with the same dimensions (height x width) as the input image.
                                                  # The dtype parameter specifies the data type of the elements in the array.

# Display the individual color channels as separate images.
# The merge function combines the zero matrices with the respective color channels to create separate images.
cv2.imshow("Red", cv2.merge([zeros, zeros, R]))     # Merge the zero matrix with the Red channel to create a Red channel image.
cv2.imshow("Green", cv2.merge([zeros, G, zeros]))   # Merge the zero matrix with the Green channel to create a Green channel image.
cv2.imshow("Blue", cv2.merge([B, zeros, zeros]))    # Merge the zero matrix with the Blue channel to create a Blue channel image.

cv2.waitKey(0)       # Wait indefinitely until a key is pressed. This is necessary to keep the OpenCV windows open.
cv2.destroyAllWindows()  # Close all OpenCV windows once the program exits the event loop.
```
