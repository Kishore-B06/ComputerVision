## Drawing images and shapes using OpenCV

Let's start off my making a black square

```python
import cv2
import numpy as np

# Create a black image
image = np.zeros((512,512,3), np.uint8) #Each pixel value is of type uint8 (unsigned 8-bit integer).

# Can we make this in black and white?
image_bw = np.zeros((512,512), np.uint8)

cv2.imshow("Black Square (Color)", image)
cv2.imshow("Black Square (B&W)", image_bw)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Let's draw a line over our black sqare

cv2.line(image, starting cordinates, ending cordinates, color, thickness)

```python
# Import the OpenCV library, providing access to image processing functions.
import cv2

# Import the NumPy library, used for numerical operations.
import numpy as np  

# Create a black image with dimensions 512x512 pixels and three color channels (BGR).
# The image is represented as a NumPy array initialized with zeros.
# Each pixel value is of type uint8 (unsigned 8-bit integer).
image = np.zeros((512,512,3), np.uint8)

# Draw a diagonal blue line on the image.
# cv2.line() function draws a line on an image.
# Parameters:
#   - image: The image on which the line will be drawn.
#   - (0,0): Starting point of the line (x, y).
#   - (511,511): Ending point of the line (x, y).
#   - (255,127,0): Color of the line in BGR format (blue in this case).
#   - 5: Thickness of the line in pixels.
cv2.line(image, (0,0), (511,511), (255,127,0), 5)

# Display the image with the diagonal blue line.
cv2.imshow("Blue Line", image)

# Wait indefinitely until a key is pressed.
# This is necessary to keep the OpenCV window open.
cv2.waitKey(0)

# Close all OpenCV windows once the program exits the event loop.
cv2.destroyAllWindows()
```

### Let's now draw a rectangle

cv2.rectangle(image, starting vertex, opposite vertex, color, thickness)

```python
# Draw a Rectangle in
image = np.zeros((512,512,3), np.uint8)

cv2.rectangle(image, (100,100), (300,250), (127,50,127), 5) #if we change "5" to -1 it will fill the image
cv2.imshow("Rectangle", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### How about cirlcles?

cv2.cirlce(image, center, radius, color, fill)

```python
image = np.zeros((512,512,3), np.uint8)

cv2.circle(image, (350, 350), 100, (15,75,50), -1) 
cv2.imshow("Circle", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## And polygons?

```python
image = np.zeros((512,512,3), np.uint8)

# Let's define four points
pts = np.array( [[10,50], [400,50], [90,200], [50,500]], np.int32)

# Let's now reshape our points in form  required by polylines
pts = pts.reshape((-1,1,2))
# Draw a polygon on the image using the specified points.
# cv2.polylines() function draws one or more polygons on an image.
# Parameters:
#   - image: The image on which the polygon(s) will be drawn.
#   - [pts]: List of polygon vertices.
#   - True: Specifies whether the polygon should be closed (True) or not (False).
#   - (0,0,255): Color of the polygon in BGR format (red in this case).
#   - 3: Thickness of the polygon edges in pixels.
cv2.polylines(image, [pts], True, (0,0,255), 3)
cv2.imshow("Polygon", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Let's even add text with cv2.putText

cv2.putText(image, 'Text to Display', bottom left starting point, Font, Font Size, Color, Thickness)

- FONT_HERSHEY_SIMPLEX, FONT_HERSHEY_PLAIN
- FONT_HERSHEY_DUPLEX,FONT_HERSHEY_COMPLEX 
- FONT_HERSHEY_TRIPLEX, FONT_HERSHEY_COMPLEX_SMALL
- FONT_HERSHEY_SCRIPT_SIMPLEX
- FONT_HERSHEY_SCRIPT_COMPLEX

```python
image = np.zeros((512,512,3), np.uint8)

# cv2.putText() function adds text to an image.
# Parameters:
#   - image: The image on which the text will be drawn.
#   - 'Hello World!': The text to be added.
#   - (75,290): The coordinates (x, y) where the text will be placed (bottom-left corner of the text).
#   - cv2.FONT_HERSHEY_COMPLEX: The font face used for the text.
#   - 2: The font scale factor that specifies the size of the text.
#   - (100,170,0): The color of the text in BGR format (in this case, a shade of green).
#   - 3: The thickness of the text strokes in pixels.
cv2.putText(image, 'Hello World!', (75,290), cv2.FONT_HERSHEY_COMPLEX, 2, (100,170,0), 3)
cv2.imshow("Hello World!", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
